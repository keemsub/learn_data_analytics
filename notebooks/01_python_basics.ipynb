```python
age = 25
height = 175.5
name = '민수'

scores = [90, 85, 88, 95]
student = {
    'name': name,
    'age': age,
    'height': height,
    'scores': scores
}

print(student)
print('평균 점수:', sum(scores) / len(scores))
```

```python
score = 87
if score >= 90:
    grade = 'A'
elif score >= 80:
    grade = 'B'
else:
    grade = 'C'

print('등급:', grade)

def average(values):
    return sum(values) / len(values)

print('평균:', average(scores))
```

## 1. 변수와 자료형

변수는 데이터를 저장하는 상자와 비슷합니다. 
예를 들어 학생의 이름, 나이, 점수 등을 변수에 담아 사용할 수 있습니다.

Python에는 여러 가지 자료형이 있습니다.

- int: 정수
- float: 실수
- str: 문자열
- bool: 참/거짓
- list: 여러 값을 담는 리스트
- dict: 키-값 쌍으로 구성된 사전

여기서 중요한 점은 각 데이터가 어떤 타입인지 아는 것입니다. 
그래야 연산을 하거나 조건문을 사용할 때 오류를 줄일 수 있습니다.

```python
# 숫자와 문자열
age = 25
height = 175.5
name = "민수"

# 리스트: 여러 값을 하나로 묶을 때 사용
scores = [90, 85, 88, 95]

# 딕셔너리: 키-값 형태로 데이터 저장
student = {
    "name": name,
    "age": age,
    "height": height,
    "scores": scores,
}

print("이름:", student["name"])
print("나이:", student["age"])
print("점수 목록:", student["scores"])
print("평균 점수:", sum(scores) / len(scores))

# 문자열 포매팅
message = f"{student['name']} 학생의 평균 점수는 {sum(scores) / len(scores):.2f}점입니다."
print(message)

# 리스트를 활용한 간단한 데이터 처리
filtered_scores = [score for score in scores if score >= 90]
print("90점 이상 점수:", filtered_scores)

# 자료형 확인
print("나이의 자료형:", type(age))
print("키의 자료형:", type(height))
print("점수 리스트의 자료형:", type(scores))
```

## 2. 조건문: 상황에 따라 다른 결과

조건문은 특정 조건이 참인지 거짓인지에 따라 다른 코드를 실행하게 해줍니다. 
예를 들어 시험 점수가 90점 이상이면 A, 80점 이상이면 B를 부여하는 로직처럼 사용할 수 있습니다.

if 조건1:
    실행할 코드
elif 조건2:
    다른 코드
else:
    기본 코드

이 구조를 잘 기억하면 프로그램이 다양한 상황을 처리할 수 있습니다.

```python
# 점수에 따라 등급을 계산하는 예시
score = 87
is_student = True

if score >= 90:
    grade = 'A'
elif score >= 80:
    grade = 'B'
elif score >= 70:
    grade = 'C'
else:
    grade = 'D'

print('점수:', score)
print('등급:', grade)

# 조건을 여러 개 조합하는 예시
is_passed = score >= 60
is_excellent = score >= 90

print('합격 여부:', is_passed)
print('우수 학생 여부:', is_excellent)

# 예시: 나이와 학생 여부에 따른 메시지
if age >= 20 and is_student:
    print('성인 학생입니다.')
else:
    print('학생이 아니거나 성인이 아닙니다.')
```

## 3. 반복문: 같은 작업을 여러 번 실행하기

반복문은 리스트의 각 원소를 순회하거나, 특정 조건이 만족될 때까지 반복할 수 있게 합니다. 
데이터 분석에서는 여러 데이터를 한 번에 처리할 때 매우 자주 사용됩니다.

- for: 정해진 범위나 리스트를 순회할 때 사용
- while: 조건이 참인 동안 계속 실행할 때 사용

예를 들어 시험 점수 목록을 순회하며 각 점수를 출력하거나 평균을 계산하는 작업이 대표적입니다.

```python
# for 문으로 리스트 순회하기
for score in scores:
    print('점수:', score)

# 반복문을 이용해 평균 점수 계산하기
total = 0
for score in scores:
    total += score

average = total / len(scores)
print('반복문으로 계산한 평균:', average)

# while 문 예시
count = 0
while count < 5:
    print('현재 count:', count)
    count += 1

# 딕셔너리 반복 예시
for key, value in student.items():
    print(f'{key}: {value}')
```

## 4. 함수: 코드를 재사용 가능하게 만들기

함수는 특정 작업을 수행하는 코드 블록입니다. 
코드를 여러 번 쓰지 않고, 필요할 때마다 호출해서 사용할 수 있습니다.

함수는 다음 구조로 만듭니다.

def 함수이름(매개변수):
    실행 코드
    return 결과

예를 들어 평균을 계산하는 함수를 만들면, 다른 데이터를 넣어도 재사용할 수 있습니다.

```python
# 함수 정의

def average(values):
    if len(values) == 0:
        return 0
    return sum(values) / len(values)


def grade_from_score(score):
    if score >= 90:
        return 'A'
    elif score >= 80:
        return 'B'
    elif score >= 70:
        return 'C'
    else:
        return 'D'


# 함수 사용
print('평균:', average(scores))
print('민수의 등급:', grade_from_score(87))

# 여러 학생의 점수에 대해 함수 적용
student_scores = {
    '민수': 92,
    '영희': 85,
    '철수': 78,
    '수지': 68
}

for name, score in student_scores.items():
    print(f'{name}: {score}점 -> {grade_from_score(score)}등급')
```

## 5. 클래스: 객체를 설계하는 방법

클래스는 비슷한 성질을 가진 여러 객체를 만들기 위한 설계도입니다. 
예를 들어 학생, 제품, 고객 같은 개념을 클래스 형태로 정의할 수 있습니다.

클래스는 다음처럼 만들 수 있습니다.

class 클래스이름:
    def __init__(self, 속성1, 속성2):
        self.속성1 = 속성1
        self.속성2 = 속성2

    def 메서드(self):
        실행 코드

클래스를 사용하면 데이터를 구조적으로 관리하고, 관련 기능도 함께 묶어서 사용할 수 있습니다.
데이터 분석에서도 CSV 데이터의 한 행, 고객 정보, 제품 정보 등을 객체로 다루는 경우가 많습니다.

```python
# 클래스 정의
class Student:
    def __init__(self, name, age, scores):
        self.name = name
        self.age = age
        self.scores = scores

    def average(self):
        if len(self.scores) == 0:
            return 0
        return sum(self.scores) / len(self.scores)

    def grade(self):
        avg = self.average()
        if avg >= 90:
            return 'A'
        elif avg >= 80:
            return 'B'
        elif avg >= 70:
            return 'C'
        else:
            return 'D'

    def intro(self):
        return f"{self.name}({self.age}세): 평균 {self.average():.2f}점, 등급 {self.grade()}"


# 인스턴스(객체) 생성
student1 = Student('민수', 25, [90, 88, 95])
student2 = Student('영희', 22, [77, 82, 91])

print(student1.intro())
print(student2.intro())
print('민수의 평균 점수:', student1.average())
print('영희의 등급:', student2.grade())
```

## 6. 예외 처리: 오류를 안전하게 다루기

프로그램을 작성할 때는 종종 예상치 못한 상황이 발생합니다. 
예를 들어 0으로 나누기, 문자열을 숫자로 변환하지 못하는 경우, 파일이 없는 경우 등이 있습니다.
이런 문제를 프로그램이 갑자기 멈추지 않도록 처리하는 방법이 바로 예외 처리입니다.

try:
    실행할 코드
except 예외종류:
    오류가 발생했을 때 실행할 코드
else:
    오류가 없을 때 실행할 코드
finally:
    마지막에 항상 실행할 코드

예외 처리를 잘 하면 사용자에게 더 친절한 메시지를 보여주고, 프로그램이 중단되지 않도록 만들 수 있습니다.

```python
# 예외 처리 예시

def divide(a, b):
    try:
        result = a / b
    except ZeroDivisionError:
        print('0으로 나눌 수 없습니다.')
        return None
    except TypeError:
        print('숫자만 입력해야 합니다.')
        return None
    else:
        print('나눗셈이 성공적으로 수행되었습니다.')
        return result
    finally:
        print('계산이 종료되었습니다.')


print('결과 1:', divide(10, 2))
print('결과 2:', divide(10, 0))
print('결과 3:', divide('10', 2))

# 사용자 입력을 안전하게 처리하는 예시
try:
    age = int(input('나이를 입력하세요: '))
    print('입력된 나이:', age)
except ValueError:
    print('숫자를 입력해야 합니다.')

# 파일 처리 예시 (실제로 존재하지 않는 파일을 열어봅니다)
try:
    with open('없는파일.txt', 'r', encoding='utf-8') as file:
        content = file.read()
        print(content)
except FileNotFoundError:
    print('파일을 찾을 수 없습니다.')
```

앞으로 강의에서도 이 내용을 계속 확장해 나갈 예정이니, 각 예제를 직접 실행해보고 이해하는 습관을 들이세요.

실습 과제:
- 리스트에 5개의 점수를 저장하고 평균을 계산해 보세요.
- 점수별로 '합격/불합격'을 출력하는 함수를 작성해 보세요.
- 학생 정보를 딕셔너리로 만들어 이름, 나이, 점수를 출력해 보세요.
- Student 클래스를 만들고, 여러 학생 객체를 생성해 보세요.
- try/except를 사용해 숫자를 입력받고 잘못된 값을 처리해 보세요.

```python
# 종합 실습 예제
students = [
    {'name': '민수', 'age': 25, 'score': 92},
    {'name': '영희', 'age': 22, 'score': 85},
    {'name': '철수', 'age': 28, 'score': 76},
    {'name': '수지', 'age': 24, 'score': 67}
]


def status(score):
    return '합격' if score >= 60 else '불합격'

for student in students:
    name = student['name']
    age = student['age']
    score = student['score']
    print(f'{name}({age}세): {score}점 -> {status(score)}')

# 한 번에 계산하는 평균
avg_score = sum(student['score'] for student in students) / len(students)
print('전체 평균 점수:', avg_score)

# 추가 연습: 90점 이상만 추려서 출력
high_scores = [student['score'] for student in students if student['score'] >= 90]
print('90점 이상 학생 점수:', high_scores)
```

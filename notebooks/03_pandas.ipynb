# pandas 기초

pandas를 이용해 표 형태 데이터를 다룹니다.

```python
import pandas as pd

students = {
    'name': ['Alice', 'Bob', 'Charlie', 'Diana'],
    'age': [23, 25, 22, 27],
    'score': [88, 92, 75, 96]
}

df = pd.DataFrame(students)
print(df)
print(df.describe())
```

```python
df['grade'] = ['A', 'A', 'C', 'A']
print(df[df['score'] >= 90])
```

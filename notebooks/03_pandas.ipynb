# pandas 기초

pandas는 표 형태 데이터(테이블)를 다루는 데 매우 강력한 라이브러리입니다. 
실무에서는 엑셀, CSV, 데이터베이스 테이블 같은 구조를 DataFrame 형태로 읽고, 필터링, 정렬, 요약, 변환을 수행합니다.

이번 노트북에서는 다음 내용을 학습합니다.

- Series와 DataFrame의 차이
- 데이터 불러오기와 기본 확인
- 열 선택과 조건 필터링
- 그룹별 집계와 정렬
- 요약 통계와 변수 생성

> 핵심 포인트: pandas는 '데이터를 분석하기 위한 표 도구'입니다. 
> 생각보다 자주 쓰이며, 데이터 분석의 거의 모든 작업에서 기본이 됩니다.

## 1. DataFrame과 Series

pandas의 가장 기본 객체는 다음과 같습니다.

- Series: 1차원 배열처럼 동작하는 데이터 구조
- DataFrame: 행과 열이 있는 표 형태의 구조

예를 들어 학생 명단은 DataFrame이고, 각 학생의 점수는 Series로 볼 수 있습니다.

이 구조를 이해하면 데이터를 선택하고 필터링하는 작업이 훨씬 쉬워집니다.

```python
import pandas as pd

students = {
    'name': ['Alice', 'Bob', 'Charlie', 'Diana'],
    'age': [23, 25, 22, 27],
    'score': [88, 92, 75, 96],
    'department': ['AI', 'Marketing', 'AI', 'Sales']
}

df = pd.DataFrame(students)
print(df)
print('\n--- 컬럼 선택 ---')
print(df['name'])
print('\n--- 요약 통계 ---')
print(df.describe())
print('\n--- 데이터 타입 확인 ---')
print(df.dtypes)
```

## 2. 기본 데이터 확인

데이터를 다룰 때 가장 먼저 해야 할 일은 데이터의 모양과 내용을 확인하는 것입니다.

- head(): 앞부분 확인
- tail(): 뒷부분 확인
- shape: 행과 열 수 확인
- info(): 컬럼 정보와 결측값 확인
- describe(): 기초 통계 확인

이 단계는 데이터가 올바르게 로드되었는지 확인하는 필수 과정입니다.

```python
import pandas as pd

# 예시 데이터 생성
sales_data = {
    'date': ['2024-01-01', '2024-01-02', '2024-01-03', '2024-01-04'],
    'region': ['서울', '부산', '서울', '대구'],
    'sales': [1200, 950, 1400, 1100],
    'customers': [110, 95, 130, 102]
}

sales_df = pd.DataFrame(sales_data)

print(sales_df.head())
print('\nshape:', sales_df.shape)
print('\ninfo:')
print(sales_df.info())
print('\n기초 통계:')
print(sales_df.describe())
```

## 3. 조건 필터링과 열 추가

실제 데이터 분석은 '조건에 맞는 자료만 골라서 보는' 과정이 많습니다. 
예를 들어 점수가 90점 이상인 학생만 선택하거나, 특정 지역 매출만 추출할 수 있습니다.

이때 `[]`, `loc`, `iloc`, `query()` 같은 방법을 사용합니다.

```python
# 학생 성적 데이터에 조건 필터링 적용
import pandas as pd

students = {
    'name': ['Alice', 'Bob', 'Charlie', 'Diana', 'Ethan'],
    'age': [23, 25, 22, 27, 24],
    'score': [88, 92, 75, 96, 84],
    'department': ['AI', 'Marketing', 'AI', 'Sales', 'AI']
}

df = pd.DataFrame(students)

# 점수 90 이상인 학생
high_score_students = df[df['score'] >= 90]
print('90점 이상 학생:')
print(high_score_students)

# 열 추가
# grade를 계산해서 넣기

def get_grade(score):
    if score >= 90:
        return 'A'
    elif score >= 80:
        return 'B'
    elif score >= 70:
        return 'C'
    else:
        return 'D'


df['grade'] = df['score'].apply(get_grade)
print('\n등급 추가 후 데이터프레임:')
print(df)

# AI 전공 학생만 필터링
ai_students = df[df['department'] == 'AI']
print('\nAI 전공 학생:')
print(ai_students)
```

## 4. 그룹별 집계와 정렬

그룹별 집계는 데이터에서 추세를 비교할 때 매우 중요합니다. 
예를 들어 지역별 매출 합계, 부서별 평균 점수, 성별 평균 나이 등을 계산할 수 있습니다.

이 작업은 pandas의 `groupby()`와 `agg()`를 사용해 수행합니다.

```python
import pandas as pd

student_df = pd.DataFrame({
    'name': ['Alice', 'Bob', 'Charlie', 'Diana', 'Ethan'],
    'department': ['AI', 'Marketing', 'AI', 'Sales', 'AI'],
    'score': [88, 92, 75, 96, 84]
})

# 부서별 평균 점수
by_department = student_df.groupby('department')['score'].mean()
print('부서별 평균 점수:')
print(by_department)

# 정렬
sorted_df = student_df.sort_values('score', ascending=False)
print('\n점수 높은 순 정렬:')
print(sorted_df)

# 학생 수 계산
count_by_department = student_df.groupby('department').size()
print('\n부서별 학생 수:')
print(count_by_department)

# 여러 집계 한 번에
summary = student_df.groupby('department').agg({
    'score': ['mean', 'max', 'min', 'count']
})
print('\n부서별 종합 요약:')
print(summary)
```

## 5. 실전 데이터 분석 흐름

pandas는 보통 다음 순서로 사용됩니다.

1. 데이터를 불러온다.
2. 기본 구조를 확인한다.
3. 결측치나 이상치를 점검한다.
4. 필요한 열을 선택한다.
5. 조건에 맞게 필터링한다.
6. 그룹별 집계를 수행한다.
7. 시각화하거나 보고용 형태로 정리한다.

이 흐름은 실제 데이터 분석 프로젝트의 거의 모든 단계에 공통으로 적용됩니다.

```python
import pandas as pd

# 실제 CSV 파일을 읽어오는 예시
sales_df = pd.read_csv('../data/sample_sales.csv')
print(sales_df.head())

# 지역별 총 판매액 계산
region_sales = sales_df.groupby('region')['sales'].sum().reset_index()
print('\n지역별 총 판매액:')
print(region_sales.sort_values('sales', ascending=False))

# 광고비와 매출의 상관관계 간단 확인
corr = sales_df[['advertising', 'sales']].corr()
print('\n광고비와 매출 상관관계:')
print(corr)

# 광고비가 200 이상인 데이터만 선택
high_ad = sales_df[sales_df['advertising'] >= 200]
print('\n광고비 200 이상 데이터 수:', len(high_ad))
```

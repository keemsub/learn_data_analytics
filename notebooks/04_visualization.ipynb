# 시각화

시각화는 데이터의 패턴, 이상치, 관계, 경향을 한눈에 보여주는 가장 강력한 도구입니다. 
숫자만 보면 잘 보이지 않던 인사이트가 그래프로 보면 훨씬 명확해집니다.

이번 노트북에서는 다음을 학습합니다.

- 막대 그래프: 범주별 비교
- 선 그래프: 시간에 따른 변화
- 산점도: 두 변수의 관계
- 히스토그램: 값의 분포
- 박스플롯: 이상치와 분포 확인

시각화는 분석 결과를 설명할 때 가장 많이 쓰이는 작업입니다. 
강의에서 단순히 코드를 외우기보다, 어떤 그래프를 어떤 상황에 쓰는지를 이해하는 것이 중요합니다.

## 1. 그래프 선택의 원리

그래프는 목적에 맞게 선택해야 합니다.

- 막대 그래프: 지역별 매출, 부서별 평균 점수와 같이 범주 비교
- 선 그래프: 월별 매출 추이처럼 시간의 흐름을 보여줄 때
- 산점도: 광고비와 매출, 온도와 판매량 같은 관계 파악
- 히스토그램: 점수 분포, 고객 연령 분포처럼 빈도 확인
- 박스플롯: 이상치와 중앙값을 빠르게 파악

차트는 단순히 예쁘게 그리는 것보다, 질문에 답을 주는 도구로 생각해야 합니다.

```python
import matplotlib
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

matplotlib.rcParams['font.family'] = ['NanumGothic', 'Malgun Gothic', 'Noto Sans CJK KR', 'AppleGothic', 'sans-serif']
matplotlib.rcParams['axes.unicode_minus'] = False

sales_df = pd.read_csv('../data/sample_sales.csv')

# 1) 지역별 총 매출 막대 그래프
region_sales = sales_df.groupby('region')['sales'].sum().sort_values()
region_sales.plot(kind='bar', color='steelblue', edgecolor='black')
plt.title('지역별 총 매출')
plt.xlabel('지역')
plt.ylabel('매출')
plt.tight_layout()
plt.show()

# 2) 시간 흐름을 보여주는 선 그래프
monthly_sales = sales_df.groupby('month')['sales'].sum().sort_values()
plt.figure(figsize=(8, 4))
plt.plot(monthly_sales.index, monthly_sales.values, marker='o', color='darkorange')
plt.title('월별 매출 추이')
plt.xlabel('월')
plt.ylabel('매출')
plt.tight_layout()
plt.show()
```

## 2. 산점도와 관계 파악

산점도는 두 변수 간의 관계를 보여줍니다. 
예를 들어 광고비가 늘수록 매출이 증가하는지, 아니면 반대로 감소하는지를 확인할 수 있습니다.

- 점들이 오른쪽 위로 올라가면 양의 관계
- 아래로 내려가면 음의 관계
- 점이 퍼져 있으면 관계가 약함

이렇게 그래프를 보면 단순한 평균보다 더 많은 정보를 얻을 수 있습니다.

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

sales_df = pd.read_csv('../data/sample_sales.csv')

# 광고비와 매출의 관계
sns.scatterplot(data=sales_df, x='advertising', y='sales', hue='region', s=100)
plt.title('광고비와 매출 관계')
plt.xlabel('광고비')
plt.ylabel('매출')
plt.tight_layout()
plt.show()

# 히스토그램: 매출 분포 확인
plt.figure(figsize=(8, 4))
plt.hist(sales_df['sales'], bins=20, color='mediumseagreen', edgecolor='black')
plt.title('매출 분포')
plt.xlabel('매출')
plt.ylabel('빈도')
plt.tight_layout()
plt.show()
```

## 3. 이상치와 분포 확인

박스플롯은 데이터의 중앙값, 사분위수, 이상치 범위를 한 번에 보여줍니다. 
특정 그룹에서 이상치가 많거나 값 분산이 큰지 빠르게 파악할 수 있습니다.

예를 들어 어떤 지역의 매출이 전체보다 훨씬 편차가 큰지 확인할 때 유용합니다.

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

sales_df = pd.read_csv('../data/sample_sales.csv')

# 박스플롯으로 지역별 매출 분포 비교
sns.boxplot(data=sales_df, x='region', y='sales', palette='Set2')
plt.title('지역별 매출 분포')
plt.xlabel('지역')
plt.ylabel('매출')
plt.tight_layout()
plt.show()

# 상관관계 시각화
sns.heatmap(sales_df[['sales', 'advertising', 'discount']].corr(), annot=True, cmap='coolwarm')
plt.title('변수 간 상관관계')
plt.tight_layout()
plt.show()
```

## 4. 시각화의 핵심 포인트

시각화는 단순히 '예쁘게 그리는 것'이 아니라, 데이터를 통해 질문에 답하는 과정입니다.

- 무엇을 비교하고 싶은지
- 어떤 패턴을 보여주려는지
- 어떤 인사이트를 전달하고 싶은지

이 세 가지를 생각하고 그래프를 그리면 훨씬 효과적인 분석이 됩니다.

예를 들어 다음 질문에 대해 서로 다른 그래프가 적합합니다.

- '지역별 매출 차이는?' -> 막대 그래프
- '월별 매출 추이는?' -> 선 그래프
- '광고비와 매출의 관계는?' -> 산점도
- '데이터 분포는 어떤가?' -> 히스토그램 또는 박스플롯

```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

sales_df = pd.read_csv('../data/sample_sales.csv')

# 실전형 종합 시각화
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

# 지역별 평균 매출
region_avg = sales_df.groupby('region')['sales'].mean().sort_values()
region_avg.plot(kind='bar', ax=axes[0], color='cornflowerblue')
axes[0].set_title('지역별 평균 매출')
axes[0].set_xlabel('지역')
axes[0].set_ylabel('평균 매출')

# 광고비와 매출 산점도
sns.scatterplot(data=sales_df, x='advertising', y='sales', hue='region', ax=axes[1], s=80)
axes[1].set_title('광고비와 매출 관계')
axes[1].set_xlabel('광고비')
axes[1].set_ylabel('매출')

plt.tight_layout()
plt.show()
```

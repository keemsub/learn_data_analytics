# 시각화

데이터의 패턴을 시각적으로 확인해 봅니다.

```python
import matplotlib
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

matplotlib.rcParams['font.family'] = ['NanumGothic', 'Malgun Gothic', 'Noto Sans CJK KR', 'AppleGothic', 'sans-serif']
matplotlib.rcParams['axes.unicode_minus'] = False

sales_df = pd.read_csv('../data/sample_sales.csv')
region_sales = sales_df.groupby('region')['sales'].sum().sort_values()
region_sales.plot(kind='bar', color='steelblue')
plt.title('지역별 총 매출')
plt.xlabel('지역')
plt.ylabel('매출')
plt.tight_layout()
plt.show()
```

```python
sns.scatterplot(data=sales_df, x='advertising', y='sales', hue='region')
plt.title('광고비와 매출 관계')
plt.tight_layout()
plt.show()
```

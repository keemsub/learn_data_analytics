# 기초 통계

통계는 데이터를 이해하고, 의미를 해석하며, 의사결정을 지원하는 핵심 도구입니다. 
데이터 분석에서는 숫자 자체보다 그 숫자가 무엇을 의미하는지를 아는 것이 중요합니다.

이번 노트북에서는 다음 개념을 살펴봅니다.

- 평균(Mean): 데이터의 중심 경향을 나타냄
- 중앙값(Median): 정렬 후 가운데 값
- 분산(Variance)과 표준편차(Standard Deviation): 데이터가 얼마나 퍼져 있는지
- 왜도, 분포, 비교 분석 등: 데이터의 형태와 특성을 이해하는 방법

실무에서는 단순히 평균만 보지 않고, 편차와 분포까지 함께 봐야 올바른 판단을 할 수 있습니다.

> 실습 팁: 아래 코드를 실행하면서 각 값이 어떤 의미를 가지는지 생각해 보세요.

## 1. 중심 경향 측정

중심 경향은 데이터가 어디에 몰려 있는지를 보여줍니다.

- 평균: 모든 값을 더한 뒤 갯수로 나눈 값
- 중앙값: 데이터를 오름차순으로 정렬했을 때 가운데 위치한 값
- 최빈값: 가장 자주 등장하는 값

예를 들어 회사의 월별 매출이 매우 높거나 낮은 달이 있으면 평균만으로는 전체 상황을 정확히 파악하기 어렵습니다. 
이런 경우 중앙값이나 분포를 함께 확인하는 것이 좋습니다.

```python
import statistics

sales = [120, 150, 170, 100, 130, 140, 160, 180, 190, 110, 145, 175]

print('매출 데이터:', sales)
print('평균:', statistics.mean(sales))
print('중앙값:', statistics.median(sales))
print('최빈값:', statistics.mode(sales))
print('최소값:', min(sales))
print('최대값:', max(sales))

# 평균이 전체 데이터의 대표값이 되는지 확인
# 예: 한 달에 큰 매출이 하나 들어가면 평균이 크게 움직일 수 있음
sales_with_outlier = [120, 150, 170, 100, 130, 140, 160, 180, 190, 110, 145, 175, 900]
print('이상치 포함 평균:', statistics.mean(sales_with_outlier))
print('이상치 포함 중앙값:', statistics.median(sales_with_outlier))
```

## 2. 분산과 표준편차

분산과 표준편차는 데이터가 얼마나 퍼져 있는지를 나타냅니다.

- 분산: 각 값이 평균으로부터 얼마나 떨어져 있는지의 평균 제곱 값
- 표준편차: 분산의 제곱근으로, 해석이 더 쉬운 값

표준편차가 크다는 것은 데이터가 넓게 퍼져 있다는 뜻입니다. 
예를 들어 판매량이 매우 일정한 제품과 변동이 큰 제품은 표준편차가 다르게 나타납니다.

```python
import statistics

sales = [120, 150, 170, 100, 130, 140, 160, 180, 190, 110, 145, 175]

mean_sales = statistics.mean(sales)
var_sales = statistics.pvariance(sales)
std_sales = statistics.pstdev(sales)

print('평균 매출:', mean_sales)
print('분산:', var_sales)
print('표준편차:', std_sales)

# 표준편차를 활용한 해석
# 평균이 145이고 표준편차가 25라면, 대체로 120~170 사이에 데이터가 몰려 있다고 볼 수 있음
print('대략적 데이터 범위(평균 ± 1표준편차):', round(mean_sales - std_sales, 2), '~', round(mean_sales + std_sales, 2))

# 두 팀의 성과 비교 예시
team_a = [80, 82, 85, 88, 90]
team_b = [60, 75, 90, 95, 100]

print('팀 A 평균:', statistics.mean(team_a), '표준편차:', statistics.pstdev(team_a))
print('팀 B 평균:', statistics.mean(team_b), '표준편차:', statistics.pstdev(team_b))

# 평균이 비슷해도 표준편차가 다르면 결과의 안정성이 다를 수 있음
```

## 3. 데이터의 분포와 해석

통계를 이해할 때 중요한 건 '평균만 보는 것'이 아니라 '분포를 보는 것'입니다.

예를 들어 A와 B의 평균 점수가 같다고 해서 두 집단이 같다고 말할 수는 없습니다. 
실제 데이터가 얼마나 퍼져 있는지, 어떤 값이 많이 모여 있는지 알아야 의미를 정확히 해석할 수 있습니다.

이런 이유로 데이터 분석에서는 반드시 시각화와 통계값을 같이 봅니다.

```python
import numpy as np

# 정규분포 형태의 점수 생성
np.random.seed(42)
score_a = np.random.normal(loc=80, scale=5, size=1000)
score_b = np.random.normal(loc=80, scale=15, size=1000)

print('A 그룹 평균:', round(score_a.mean(), 2), '표준편차:', round(score_a.std(), 2))
print('B 그룹 평균:', round(score_b.mean(), 2), '표준편차:', round(score_b.std(), 2))

# 평균은 비슷해도 분산이 크게 다름
# 즉, B 그룹은 변동성이 크고 불안정하다는 해석이 가능함

# 간단한 설명용 비율 계산
low_score_ratio = np.mean(score_b < 60)
print('B 그룹에서 60점 미만 비율:', round(low_score_ratio, 4))
```

## 4. 정리

통계는 그냥 숫자를 계산하는 도구가 아니라, 데이터를 이해하는 언어입니다.

- 평균은 중심 경향을 보여줍니다.
- 중앙값은 이상치에 덜 민감합니다.
- 분산과 표준편차는 데이터의 변동성을 보여줍니다.
- 분포를 함께 살펴야 의미 있는 해석이 가능합니다.

데이터 분석에서는 이 기본 통계값이 모든 작업의 시작점이 됩니다. 
다음 단계에서는 pandas를 사용해 실제 표 데이터에서 이런 통계값을 어떻게 계산하는지 배우겠습니다.

```python
# 종합 예제: 판매 데이터 요약
sales = [100, 120, 130, 110, 150, 160, 180, 140, 200, 210]

summary = {
    '평균': statistics.mean(sales),
    '중앙값': statistics.median(sales),
    '최소값': min(sales),
    '최대값': max(sales),
    '표준편차': statistics.pstdev(sales)
}

for key, value in summary.items():
    print(f'{key}: {value}')

print('해석: 평균 매출은 약', round(summary['평균'], 2), '이며, 표준편차가 크지 않아서 상대적으로 안정적이라고 볼 수 있습니다.')
```

# 종합 프로젝트

마지막으로 판매 데이터로 간단한 분석을 수행해 봅니다.

```python
import pandas as pd
import matplotlib.pyplot as plt

sales_df = pd.read_csv('../data/sample_sales.csv')
avg_by_region = sales_df.groupby('region', as_index=False)['sales'].mean()
print(avg_by_region)

plt.bar(avg_by_region['region'], avg_by_region['sales'])
plt.title('지역별 평균 매출')
plt.xlabel('지역')
plt.ylabel('평균 매출')
plt.tight_layout()
plt.show()
```

```python
corr = sales_df['advertising'].corr(sales_df['sales'])
print('광고비와 매출 상관계수:', corr)
```
